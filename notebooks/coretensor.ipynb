```python
# import scipy as sp
import numpy as np
import scipy.stats as stats
import scipy.linalg as la
# import matplotlib.pyplot as plt

import tensorly as tl
```

Get $SL(2, \mathbb{C})$

```python
# generators from sl2 lie algebra

e = np.array([
    [0, 1],
    [0, 0]
])
f = np.array([
    [0, 0],
    [1, 0]
])
h = np.array([
    [1, 0],
    [0, -1]
])

# some SL2

# A in SL_2
def form_sl2(α, β, γ):
    return la.expm(α * + e + β * f + γ * h) 

# get vector in W orbit
W = W.reshape((2, 2, 2))
W
```

modal product W with $SL_2$

```python
A = form_sl2(1, 0, 0)
B = form_sl2(0, 1, 0)
C = form_sl2(0, 0, 1)
```

```python
tl.tenalg.multi_mode_dot(W, [A, B, C], modes=[0, 1, 2])
```

```python
modal_prod = tl.tenalg.multi_mode_dot(W, [A], modes=[0, 1, 2])
modal_prod
```

```python
tucker_decomp = tl.decomposition.tucker(modal_prod, 2)
core_tensor, factors = tucker_decomp#.core, tucker_decomp.factors

core_tensor
```

```python
modal_prod = tl.tenalg.multi_mode_dot(W, [B], modes=[0, 1, 2])
modal_prod

tucker_decomp = tl.decomposition.tucker(modal_prod, 2)
core_tensor, factors = tucker_decomp#.core, tucker_decomp.factors

core_tensor
```

```python
modal_prod = tl.tenalg.multi_mode_dot(W, [C], modes=[0, 1, 2])
modal_prod

tucker_decomp = tl.decomposition.tucker(modal_prod, 2)
core_tensor, factors = tucker_decomp#.core, tucker_decomp.factors

core_tensor
```
